## A traffic system calculates fines:

Speed limit = 60 km/h
* If speed ≤ 60 → No fine
* If 61–80 → Fine = 500
* If 81–100 → Fine = 2000
* If >100 → License suspended
#### BUT:
*If emergency vehicle → no fine
*If rainy weather AND speed > 60 → fine increases by 50%

```python
speed=float(input("Enter speed (km/h): ")); emergency=input("Emergency vehicle (yes/no): ").lower(); rainy=input("Rainy weather (yes/no): ").lower()
if emergency=="yes" or speed<=60: print("No fine")
elif speed<=80:
    fine=500
    if rainy=="yes": fine*=1.5
    print("Fine =",fine)
elif speed<=100:
    fine=2000
    if rainy=="yes": fine*=1.5
    print("Fine =",fine)
else: print("License suspended")
```

## Write a program to decide passenger clearance status.
Take input:
* baggage_weight
* passport_valid (Yes/No)
* visa_status (Valid/Invalid)
* criminal_check (Clear/Flagged)
#### Rules:
* baggage > 30kg → extra charge
* passport invalid → reject
* visa invalid → reject
* criminal flagged → security hold
* otherwise → cleared for boarding

```python
weight=float(input("Enter baggage weight (kg): ")); passport=input("Passport valid (Yes/No): ").lower(); visa=input("Visa status (Valid/Invalid): ").lower(); criminal=input("Criminal check (Clear/Flagged): ").lower()
if passport=="no": print("Reject")
elif visa=="invalid": print("Reject")
elif criminal=="flagged": print("Security hold")
elif weight>30: print("Extra charge")
else: print("Cleared for boarding")
```

## Write a program to detect suspicious banking transactions.
Take input:
* transaction_amount
* location_match (Yes/No)
* time_of_transaction(Day/Night)
* previous_transactions_count
* device_verified (Yes/No)
#### Rules:
* amount > 100,000 → suspicious
* location mismatch + high amount → fraud alert
* night transaction + unverified device → high risk
* many transactions (>10 in 1 hour) → freeze account
* otherwise → normal transaction

```python
amount=float(input("Enter transaction amount: ")); location=input("Location match (Yes/No): ").lower(); time=input("Time of transaction (Day/Night): ").lower(); count=int(input("Previous transactions count: ")); device=input("Device verified (Yes/No): ").lower()
if count>10: print("Freeze account")
elif location=="no" and amount>100000: print("Fraud alert")
elif time=="night" and device=="no": print("High risk")
elif amount>100000: print("Suspicious")
else: print("Normal transaction")
```

## Write a program to classify a person’s risk level in a police database.
Take input:
* criminal_record (Yes/No)
* number_of_cases
* travel_history_flagged (Yes/No)
* financial_activity_score
* last_activity_days
#### Rules:
* criminal_record = Yes AND cases > 3 → High Risk
* flagged travel + suspicious financial score → High Risk
* no activity for long time (>365 days) → Monitor
* clean record + normal activity → Low Risk
* mixed signals → Medium Risk

```python
criminal=input("Criminal record (Yes/No): ").lower(); cases=int(input("Number of cases: ")); travel=input("Travel history flagged (Yes/No): ").lower(); score=float(input("Financial activity score: ")); days=int(input("Days since last activity: "))
if criminal=="yes" and cases>3: print("High Risk")
elif travel=="yes" and score>=70: print("High Risk")
elif days>365: print("Monitor")
elif criminal=="no" and travel=="no" and score<70: print("Low Risk")
else: print("Medium Risk")
```
