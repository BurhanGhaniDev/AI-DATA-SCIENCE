### Q14. Weak Password Rejection

A website requires passwords to contain at least 8 characters, including at least one digit. A password that does not meet the rules must be rejected.

Create a custom exception named `WeakPasswordError`. Ask the user for a password, check both rules, raise the custom exception when a rule fails, and handle it to explain what needs to be corrected. Do not display the actual password in the success message.

```python
class WeakPasswordError(Exception):
    pass

try:
    password = input("Create a password: ")
    if len(password) < 8 or not any(ch.isdigit() for ch in password):
        raise WeakPasswordError("Password needs at least 8 characters and one digit.")
except WeakPasswordError as e:
    print("Password rejected:", e)
else:
    print("Password accepted.")
```

### Q20. Library Borrowing Rules

A library member may borrow up to three books at a time. A member who already has three borrowed books cannot borrow another one. The requested number of books must be a positive integer.

Create custom exceptions named `BorrowLimitExceededError` and `InvalidBookCountError`. Validate the request, raise the correct exception, handle each case separately, and update the member's borrowed-book count only after a valid request.

### Q21. ATM with Multiple Custom Errors

Build a small ATM withdrawal program. The withdrawal must be greater than zero, must not exceed the account balance, and must not exceed a daily withdrawal limit.

Create custom exception classes named `InvalidAmountError`, `InsufficientBalanceError`, and `DailyLimitExceededError`. Ask for a withdrawal amount and validate each rule. Raise the relevant custom exception and handle each one with a different message. Update the balance and daily total only after all validations pass.

```python
class InvalidAmountError(Exception):
    pass

class InsufficientBalanceError(Exception):
    pass

class DailyLimitExceededError(Exception):
    pass

balance = 20000
daily_limit = 10000
withdrawn_today = 2000
try:
    amount = float(input("Enter withdrawal amount: "))
    if amount <= 0:
        raise InvalidAmountError("Amount must be positive.")
    if amount > balance:
        raise InsufficientBalanceError("Insufficient balance.")
    if withdrawn_today + amount > daily_limit:
        raise DailyLimitExceededError("Daily withdrawal limit exceeded.")
except ValueError:
    print("Enter a numeric amount.")
except (InvalidAmountError, InsufficientBalanceError, DailyLimitExceededError) as e:
    print("Withdrawal failed:", e)
else:
    balance -= amount
    withdrawn_today += amount
    print(f"New balance: Rs. {balance:.2f}")
    print(f"Withdrawn today: Rs. {withdrawn_today:.2f}")
```

### Q22. Bank Transfer Between Two Accounts

A banking system transfers money from one account to another. The transfer amount must be positive, the sender and receiver must be different accounts, and the sender must have enough funds.

Create custom exceptions named `InvalidTransferError` and `InsufficientFundsError`. Validate all rules and raise the correct exception. Handle the exceptions and ensure that neither account balance changes when any validation fails. Display both balances after the transaction attempt.

```python
class InvalidTransferError(Exception):
    pass

class InsufficientFundsError(Exception):
    pass

account_a = {"name": "A", "balance": 15000.0}
account_b = {"name": "B", "balance": 5000.0}
try:
    amount = float(input("Transfer amount from A to B: "))
    if amount <= 0:
        raise InvalidTransferError("Transfer amount must be positive.")
    if account_a is account_b:
        raise InvalidTransferError("Accounts must be different.")
    if amount > account_a["balance"]:
        raise InsufficientFundsError("Not enough funds in account A.")
except ValueError:
    print("Enter a valid numeric amount.")
except (InvalidTransferError, InsufficientFundsError) as e:
    print("Transfer failed:", e)
else:
    account_a["balance"] -= amount
    account_b["balance"] += amount
    print("Transfer successful.")
finally:
    print("Account A:", account_a["balance"])
    print("Account B:", account_b["balance"])
```

### Q24. Flight Seat Reservation

An airline has a fixed number of seats. A passenger can reserve one or more seats, but the request cannot be zero, negative, or greater than the remaining seats.

Create custom exceptions named `InvalidSeatRequestError` and `SeatsUnavailableError`. Validate user input, raise and handle the appropriate custom exception, and update the remaining-seat count only when the reservation succeeds. Test a valid booking and at least two invalid booking attempts.

```python
class InvalidSeatRequestError(Exception):
    pass

class SeatsUnavailableError(Exception):
    pass

remaining_seats = 20

def reserve_seats(request):
    global remaining_seats
    try:
        if request <= 0:
            raise InvalidSeatRequestError("Request must be at least one seat.")
        if request > remaining_seats:
            raise SeatsUnavailableError("Not enough seats available.")
    except (InvalidSeatRequestError, SeatsUnavailableError) as e:
        print("Reservation failed:", e)
    else:
        remaining_seats -= request
        print(f"Reserved {request} seat(s).")

reserve_seats(3)
reserve_seats(0)
reserve_seats(25)
print("Seats remaining:", remaining_seats)
```

### Q28. Bulk Employee Payroll Processing

A company stores several employee records in a list. Each record contains an employee name, hours worked, and hourly rate. One invalid record should not stop the payroll calculation for all other employees.

Create a custom exception named `InvalidPayrollRecordError`. Validate each employee record, raise this exception when required data is missing or values are invalid, and handle the error inside the record-processing loop. Calculate payroll for valid employees and report which records were skipped and why.

```python
class InvalidPayrollRecordError(Exception):
    pass

employees = [
    {"name": "Ali", "hours": 40, "rate": 500},
    {"name": "Sara", "hours": 35, "rate": 600},
    {"name": "", "hours": 30, "rate": 450},
    {"name": "Ahmed", "hours": -5, "rate": 400},
    {"name": "Zara", "hours": 45, "rate": 0}
]
total_payroll = 0
for employee in employees:
    try:
        name, hours, rate = employee["name"], employee["hours"], employee["rate"]
        if not name:
            raise InvalidPayrollRecordError("Employee name is missing.")
        if not 0 <= hours <= 80 or rate <= 0:
            raise InvalidPayrollRecordError("Hours must be 0-80 and rate must be positive.")
        pay = hours * rate
    except (KeyError, TypeError, InvalidPayrollRecordError) as e:
        print("Skipped invalid record:", e)
    else:
        total_payroll += pay
        print(f"{name}: Rs. {pay:.2f}")
print(f"Total payroll: Rs. {total_payroll:.2f}")
```

### Q29. Inventory Purchase with Safe Transaction

A warehouse purchase operation receives a product ID and requested quantity. The product must exist, the quantity must be positive, and enough stock must be available. A failed purchase must leave inventory unchanged.

Create custom exceptions named `ProductNotFoundError`, `InvalidQuantityError`, and `InsufficientStockError`. Implement validation before changing stock. Use `try`, `except`, `else`, and `finally` meaningfully: report success only in `else`, handle expected failures in `except`, and display an audit/attempt-finished message in `finally`. Demonstrate that stock remains unchanged after a failed purchase.

```python
class ProductNotFoundError(Exception):
    pass

class InvalidQuantityError(Exception):
    pass

class InsufficientStockError(Exception):
    pass

inventory = {"P101": 10, "P102": 5, "P103": 0}
try:
    product_id = input("Enter product ID: ").strip().upper()
    quantity = int(input("Enter quantity: "))
    if product_id not in inventory:
        raise ProductNotFoundError("Product does not exist.")
    if quantity <= 0:
        raise InvalidQuantityError("Quantity must be positive.")
    if quantity > inventory[product_id]:
        raise InsufficientStockError("Not enough stock.")
except ValueError:
    print("Enter a whole-number quantity.")
except (ProductNotFoundError, InvalidQuantityError, InsufficientStockError) as e:
    print("Purchase failed:", e)
else:
    inventory[product_id] -= quantity
    print("Purchase successful. Updated inventory:", inventory)
finally:
    print("Purchase attempt completed.")
```

### Q30. Mini Banking System with a Custom Exception Hierarchy

Design a menu-based mini banking program that supports deposit, withdrawal, and balance viewing. Deposits and withdrawals must be positive. A withdrawal cannot exceed the available balance. Invalid menu choices must also be handled.

Create a base custom exception named `BankingError`, then create `InvalidAmountError` and `InsufficientBalanceError` as child classes of `BankingError`. Raise the appropriate exception from the relevant operation and handle the base class or specific child classes safely. Keep the program running after an invalid transaction, and show the final account balance when the user chooses to exit.

```python
class BankingError(Exception):
    pass

class InvalidAmountError(BankingError):
    pass

class InsufficientBalanceError(BankingError):
    pass

balance = 10000.0
while True:
    print("\n1. Deposit  2. Withdraw  3. View balance  4. Exit")
    choice = input("Choose (1-4): ").strip()
    try:
        if choice == "1":
            amount = float(input("Deposit amount: "))
            if amount <= 0:
                raise InvalidAmountError("Amount must be positive.")
            balance += amount
            print(f"Deposit successful. Balance: Rs. {balance:.2f}")
        elif choice == "2":
            amount = float(input("Withdrawal amount: "))
            if amount <= 0:
                raise InvalidAmountError("Amount must be positive.")
            if amount > balance:
                raise InsufficientBalanceError("Insufficient balance.")
            balance -= amount
            print(f"Withdrawal successful. Balance: Rs. {balance:.2f}")
        elif choice == "3":
            print(f"Current balance: Rs. {balance:.2f}")
        elif choice == "4":
            print("Thank you for using the banking system.")
            break
        else:
            print("Choose a valid menu option.")
    except ValueError:
        print("Please enter a valid number.")
    except BankingError as e:
        print("Transaction failed:", e)
```
