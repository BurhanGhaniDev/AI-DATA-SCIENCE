```python
n = int(input("Enter a non-negative integer: "))
if n < 0:
    print("Factorial is not defined for negative integers.")
else:
    factorial = 1
    for i in range(1, n + 1):
        factorial *= i
    print("Factorial =", factorial)
```

## Student Marks Entry

### Goal:
    Record student marks and skip invalid entries.

### Steps:

    Create an empty list marks = [].
    
    Ask user to enter marks for 5 students.
    
    If marks < 0 or > 100 → print “Invalid marks” and continue.
    
    If user enters -1 → stop taking input (break).
    
    Print valid marks and their average.

### Example:

    Enter marks: 90  
    Enter marks: 105 → Invalid marks  
    Enter marks: 80  
    Enter marks: -1 → stop  
    Valid Marks: [90, 80]  
    Average = 85.0

```python
marks = []
for i in range(5):
    entry = input(f"Enter marks for student {i + 1} (or -1 to stop): ").strip()
    if entry == "-1":
        break
    try:
        mark = float(entry)
    except ValueError:
        print("Invalid marks")
        continue
    if mark < 0 or mark > 100:
        print("Invalid marks")
        continue
    marks.append(mark)

print("Valid Marks:", marks)
if marks:
    print("Average =", sum(marks) / len(marks))
else:
    print("No valid marks entered.")
```

## 🏨 8. Hotel Room Booking

### Goal:
    Manage simple room booking.

### Task:

    Create a list of rooms [101, 102, 103, 104, 105]
    
    Ask user to choose a room.
    
    If room already booked, print “Room not available.”
    
    Continue until all rooms are booked.

### Example:

    Available rooms: [101, 102, 103, 104, 105]
    Choose room: 103
    Room 103 booked successfully!

```python
available_rooms = [101, 102, 103, 104, 105]
booked_rooms = []
while available_rooms:
    print("Available rooms:", available_rooms)
    try:
        room = int(input("Choose a room: "))
    except ValueError:
        print("Please enter a valid room number.")
        continue
    if room not in available_rooms:
        print("Room not available.")
        continue
    available_rooms.remove(room)
    booked_rooms.append(room)
    print(f"Room {room} booked successfully!")
print("All rooms are booked.")
print("Booked rooms:", booked_rooms)
```

## Create a menu-driven banking system using loops:
* Create Account
* Deposit
* Withdraw
* Check Balance
* Exit

```python
balance = 0.0
while True:
    print("\n--- Banking Menu ---")
    print("1. Create Account")
    print("2. Deposit")
    print("3. Withdraw")
    print("4. Check Balance")
    print("5. Exit")
    choice = input("Choose an option: ").strip()

    if choice == "1":
        print("Account is ready to use.")
    elif choice == "2":
        amount = float(input("Enter deposit amount: "))
        if amount > 0:
            balance += amount
            print("Deposit successful.")
        else:
            print("Amount must be positive.")
    elif choice == "3":
        amount = float(input("Enter withdrawal amount: "))
        if amount <= 0:
            print("Amount must be positive.")
        elif amount > balance:
            print("Insufficient balance.")
        else:
            balance -= amount
            print("Withdrawal successful.")
    elif choice == "4":
        print("Current balance =", balance)
    elif choice == "5":
        print("Thank you for using the banking system.")
        break
    else:
        print("Invalid option.")
```

## Invoice Tax Calculator
### Goal:

Take multiple items in multiple invoices and calculate tax per invoice.

### Steps:

Outer loop → for 3 invoices.

Inner loop → 4 items each.

Input item price, add to list.

Calculate total + 10% tax.

Print invoice total.

### Example:
    Invoice 1:
    Items: [100, 200, 300, 400]
    Invoice 2:
    Items: [250, 150, 350, 250]
    → Totals: [1100, 1100]

```python
invoice_totals = []
for invoice in range(1, 4):
    subtotal = 0
    print(f"Invoice {invoice}:")
    for item in range(1, 5):
        price = float(input(f"Item {item} price: "))
        if price < 0:
            print("Price cannot be negative; treating it as 0.")
            price = 0
        subtotal += price
    tax = subtotal * 0.10
    total = subtotal + tax
    invoice_totals.append(total)
    print(f"Subtotal = {subtotal:.2f}, Tax = {tax:.2f}, Total = {total:.2f}")
print("Invoice totals:", invoice_totals)
```

## Student Result Sheet
### Goal:

    Generate report cards for each student in a class.

### Steps:

    Outer loop → 5 students.
    
    Inner loop → 4 subjects.

    If marks < 33 → print “Failed in Subject”.

    Calculate total and grade per student.

### Example:
    Student 1: [80, 90, 70, 85] → Grade: A  
    Student 2: [50, 32, 45, 40] → Failed in Subject 2

```python
for student in range(1, 6):
    marks = []
    print(f"Student {student}:")
    for subject in range(1, 5):
        mark = float(input(f"Marks for subject {subject}: "))
        marks.append(mark)
        if mark < 33:
            print(f"Failed in Subject {subject}")
    total = sum(marks)
    percentage = total / 4
    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"
    if any(mark < 33 for mark in marks):
        result = "Fail"
    else:
        result = "Pass"
    print(f"Marks: {marks}, Total: {total}, Percentage: {percentage:.2f}%, Grade: {grade}, Result: {result}")
```

```python
packages = []
for package in range(1, 4):
    ratings = []
    print(f"Package {package}:")
    for feature in range(1, 4):
        rating = float(input(f"Rating for feature {feature} (0-5): "))
        while rating < 0 or rating > 5:
            print("Rating must be between 0 and 5.")
            rating = float(input(f"Rating for feature {feature} (0-5): "))
        if rating == 0:
            continue
        ratings.append(rating)
    average = sum(ratings) / len(ratings) if ratings else 0
    packages.append((ratings, average))
    print(f"Ratings: {ratings}, Average = {average:.2f}")
```
